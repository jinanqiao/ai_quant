```python
# 实现调用deepseek的功能
def call_deepseek(prompt,history,model):
    history.append({"role": "user", "content": prompt})
    client = OpenAI(api_key=ds_apikey, base_url=ds_url)

    response = client.chat.completions.create(
        model=model,
        messages=history
    )

    reply = response.choices[0].message.content
    history.append({"role": "assistant", "content": reply})

    return reply
```

```python
# 定义第一个prompt：分析财报内容

report = """\
苹果公司2024财年Q2财报摘要：
- 总营收：907.5亿美元，同比下降4.3%（连续第五季度下滑）；
- 净利润：236.4亿美元，同比下降5.1%；
- 分业务营收：
  - iPhone：459.6亿美元，同比下降10.7%（中国市场竞争加剧）；
  - 服务（Services）：239.8亿美元（历史新高），同比增长12.3%；
  - Mac/MacBook：74.5亿美元，同比增长3.2%（M3芯片新品推动）；
  - iPad：56.4亿美元，同比下降17%；
- 毛利率：44.6%（同比微升，因服务业务占比提高）；
- 现金储备：1620亿美元（环比减少3%）。
"""

response = call_deepseek("请帮我分析财报内容,限制500个字："+report, history, ds_model)
response
```
