# yfinance Ticker 常用方法说明

## 📊 分红与拆股相关

- **`t.dividends`** - 历史股息记录（时间序列，每股分红金额）
- **`t.splits`** - 历史拆股记录（时间序列，拆股比例，如 2.0 = 2:1）
- **`t.actions`** - 公司行为汇总（包含分红、拆股等）
- **`t.capital_gains`** - 资本利得分配（主要用于 ETF/基金，普通股票通常为空）

## 📈 基本信息

- **`t.info`** - 公司详细信息（字典，100+ 字段：财务指标、公司概况、估值等）
- **`t.fast_info`** - 快速信息（延迟加载字典，约 20 个常用字段：价格、市值、均线等）
- **`t.news`** - 相关新闻（列表，最近新闻文章，包含标题、摘要、链接等）

## 💰 财务报表

- **`t.income_stmt`** - 利润表（年度，DataFrame，营收、成本、净利润等）
- **`t.quarterly_income_stmt`** - 季度利润表（季度，DataFrame）
- **`t.ttm_income_stmt`** - 过去 12 个月利润表（TTM，DataFrame）
- **`t.balance_sheet`** - 资产负债表（年度，DataFrame，资产、负债、权益等）
- **`t.cashflow`** - 现金流量表（年度，DataFrame，经营/投资/筹资现金流）
- **`t.quarterly_cashflow`** - 季度现金流量表（季度，DataFrame）
- **`t.ttm_cashflow`** - 过去 12 个月现金流量表（TTM，DataFrame）
- **`t.earnings`** - 盈利数据（年度/季度，DataFrame，EPS 等）

## 📅 事件与文件

- **`t.calendar`** - 公司事件日历（字典：财报日期、EPS 预期、分红日等）
- **`t.earnings_dates`** - 财报发布日期（DataFrame：历史及预期财报日期、EPS 预期/实际、超预期幅度）
- **`t.sec_filings`** - SEC 监管文件（字典列表：10-K、10-Q、8-K 等文件及链接）

---

**使用提示：**
- 📅 **年度数据**：默认返回年度数据
- 📊 **季度数据**：使用 `quarterly_` 前缀
- 🔄 **TTM 数据**：使用 `ttm_` 前缀（Trailing Twelve Months，过去 12 个月）
- 📋 **数据格式**：财务报表返回 DataFrame，基本信息返回字典，新闻返回列表

```python
# 18. SEC 监管文件（10-K、10-Q、8-K 等文件及链接）
sec_filings = t.sec_filings
print(f"共获取到 {len(sec_filings)} 个 SEC 文件\n")

# 显示最近5个文件
print("=== 最近5个 SEC 文件 ===")
for i, filing in enumerate(sec_filings[:5], 1):
    date_value = filing.get('date')
    
    # 处理日期：可能是时间戳（整数）或日期对象
    if isinstance(date_value, (int, float)):
        from datetime import datetime
        date_str = datetime.fromtimestamp(date_value).strftime('%Y-%m-%d')
    elif hasattr(date_value, 'strftime'):
        date_str = date_value.strftime('%Y-%m-%d')
    else:
        date_str = str(date_value)
    
    print(f"{i}. {filing.get('type', 'Unknown')} - {filing.get('title', '无标题')}")
    print(f"   日期: {date_str}")
    print(f"   链接: {filing.get('edgarUrl', '无链接')}\n")
```
