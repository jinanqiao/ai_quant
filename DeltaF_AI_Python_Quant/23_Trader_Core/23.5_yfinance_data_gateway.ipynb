```python
# 2. 调用 yfinance_gateway 模拟多消费者使用场景
from deltafq.live.event_engine import EventEngine, EVENT_TICK
from deltafq.live.gateway_registry import create_data_gateway
```

```python
# 2.1 定义消费者：2个消费渠道

def consumer_1(tick):
    print(f"前端展示界面：{tick.symbol}, 最新价格：{tick.price}")

def consumer_2(tick):
    print(f"撮合pending状态的委托单：{tick.symbol}, 最新价格：{tick.price}, 交易成功！")
```

```python
# 2.2 定义生产者，标准化流程：gateway -> set_tick_handler -> event_engine.on(消费)
import time

# 消费实现
event_engine = EventEngine()
event_engine.on(EVENT_TICK, consumer_1)
event_engine.on(EVENT_TICK, consumer_2)

# 生产实现：推送行情数据
gateway = create_data_gateway("yfinance", interval=5)
gateway.set_tick_handler(lambda tick: event_engine.emit(EVENT_TICK, tick))

# 启动实时数据拉取
if gateway.connect():
    gateway.start()
    symbols = ["BTC-USD"]  # 全天候投资标的
    gateway.subscribe(symbols)
    try:
        while True:
            time.sleep(1)
    except KeyboardInterrupt:
        gateway.stop()
```
