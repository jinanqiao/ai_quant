```python
import yfinance as yf
import time
from datetime import datetime
```

```python
# 实例1：单次获取
symbol = "000001.SS"  # 可改为 000001.SS（上证）、GOOGL、BTC-USD 等
ticker = yf.Ticker(symbol)
info = ticker.fast_info
info
```

```python
info.open
```

```python
price = info.last_price      # 最新价
volume = info.last_volume    # 最新成交量（可能为 None）
high = info.day_high
low = info.day_low
print(f"{symbol} 最新价: {price}, 成交量: {volume}, 最低价：{high}, 最高价：{low}")
```

```python
# 实例2：轮询获取
symbols = ["000001.SS"]
interval = 5  # 秒

while True:
    for symbol in symbols:
        try:
            ticker = yf.Ticker(symbol)
            info = ticker.fast_info
            price = info.last_price
            volume = info.last_volume or 0
            print(f"[{datetime.now().strftime('%H:%M:%S')}] {symbol} -> {price} | Vol: {volume}")
        except Exception as e:
            print(f"{symbol} 获取失败: {e}")
    time.sleep(interval)
```

```python
# 实例3：获取日内历史分钟数据
symbol3 = ["BTC-USD"]
data3 = yf.download(symbol3, period="1d", interval="1m", progress=False)
data3
```
