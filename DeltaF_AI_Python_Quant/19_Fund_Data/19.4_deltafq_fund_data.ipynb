```python
import deltafq as dfq
import pandas as pd
```

```python
# 做一个简单的示例，确保方法可以正常运行
fetcher = dfq.data.DataFetcher()
data = fetcher.fetch_fund_data(code="000001", page=1)
data
```

```python
# 获取自选基金：全球指数基金
fund_list = ["050025", "270042", "160119", "510300", "000071", "510880"]

for code in fund_list:
    print(f'\n==========正在下载基金历史数据：{code}')
    data = fetcher.fetch_fund_data(code=code, page=None) # 获取历史净值
    data['净值日期'] = pd.to_datetime(data['净值日期']) # 日期格式化
    data = data.sort_values('净值日期').reset_index(drop=True) # 日期正序、设置为索引
    data.to_csv(f'{code}.csv',index=False)
    print(data.head())
    print(f'已下载并保存完成：{code}')
```
