```python
def add_stock(stock_code, name, prices):
    stocks[stock_code] = [name, prices]
    print(f"已添加股票：{name}")

add_stock("03690", "美团", [120.5, 122.3, 118.7])
stocks
```

```python
# 任务2：查找最新股价
```

```python
def get_latest_price(stock_code):
    if stock_code in stocks:
        latest_price = stocks[stock_code][1][-1]
        return f"{stocks[stock_code][0]}的最新股价：${latest_price:.2f}"
    return "股票代码不存在"

get_latest_price("AAPL")
```

```python
# 任务3：计算单日涨跌幅
```

```python
def calculate_daily_change(stock_code):
    if stock_code in stocks and len(stocks[stock_code][1]) >= 2:
        prev_price = stocks[stock_code][1][-2]
        curr_price = stocks[stock_code][1][-1]
        change = (curr_price - prev_price) / prev_price * 100
        return f"{stocks[stock_code][0]}单日涨跌幅：{change:.1f}%"
    return "数据不足"

calculate_daily_change("AAPL")
```
