```python
# 选取最新的100条数据，仅保留开盘价和收盘价
```

```python
df_100 = df.tail(100)[['Open','Close']]
df_100
```

```python
# 对其 df_100 计算每日涨跌幅，并保存至新增列 -- (close - open) / open * 100
```

```python
df_100['Rate'] = (df_100['Close'] - df_100['Open'])/df_100['Open']*100
df_100
```

```python
df_100.round(2)
```
