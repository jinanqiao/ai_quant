```python
# 5. 基金数据管理练习
# 初始化基金数据库（字典存储基金信息，列表存储每日净值）
fund_db = {
    "001": ["华夏成长", [1.52, 1.55, 1.53, 1.58]],  # 格式：基金代码: [基金名称, [净值列表]]
    "002": ["天弘债券", [1.12, 1.13, 1.12, 1.14]],
    "003": ["易方达消费", [2.45, 2.50, 2.48, 2.55]]
}
```

```python
fund_db
```

```python
# 5.1：添加新基金
```

```python
fund_db["004"]=["delta基金", [1,2,3,4]]
fund_db
```

```python
# 5.2：查询最新价格
```

```python
fund_db["002"]
```

```python
fund_db["002"][1]
```

```python
fund_db["002"][1][-1]
```

```python
def get_latest_price(fund_code):
    return fund_db[fund_code][1][-1]
```

```python
abc = get_latest_price("004")
abc
```

```python
# 5.3：计算基金收益率：(最新价格 - 初始价格) / 初始价格 * 100  -- 格式化字符串 f"收益率：{rate:.1f}%"
```

```python
price2 = fund_db["002"][1][-1]
price1 = fund_db["002"][1][0]
price2,price1
```

```python
rate = (price2 - price1) / price1 * 100
rate
```

```python
def calculate_rate(fund_code):
    price2 = fund_db[fund_code][1][-1]
    price1 = fund_db[fund_code][1][0]
    rate = (price2 - price1) / price1 * 100
    return f"收益率：{rate:.2f}%"
```

```python
calculate_rate("002")
```
